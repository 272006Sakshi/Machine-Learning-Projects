```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2-d line plot: 
- it is used for bivariate analysis
- categorical -> numerical  and  numerical->numerical
- usecase: time-series data

```python
#plotting 
price = [30000, 35000, 38500, 42000, 35600, 42000, 40000]
year = [2013, 2015, 2016, 2017, 2020, 2023, 2025]

plt.plot(year, price)
```

```python
data = pd.read_csv("C:\\Users\\HP\\Downloads\\sharma-kohli.csv")
```

```python
data
```

```python
plt.plot(data['index'], data['RG Sharma'])
```

```python
plt.plot(data['index'], data['V Kohli'])
```

```python
# both in single plot
plt.plot(data['index'], data['RG Sharma'], color='blue', linestyle='dashed', label='Rohit sharma')
plt.plot(data['index'], data['V Kohli'], color='red', linestyle='dashed', label='Virat Kohli')
# add title
plt.title("Rohit sharma Vs Virat Kohli career comparison")
#add leble
plt.xlabel('season')
plt.ylabel('Score')

# to show label we use legend function
plt.legend()
```

```python
price = [30000, 35000, 38500, 42000, 35600, 42000, 500000]
year = [2013, 2015, 2016, 2017, 2020, 2023, 2025]

plt.plot(year, price)
# add y limit
plt.ylim(0, 50000)
```

```python
price = [30000, 35000, 38500, 42000, 35600, 42000, 500000]
year = [2013, 2015, 2016, 2017, 2020, 2023, 300000]

plt.plot(year, price)
# similiary here also we can add limit
plt.ylim(20000, 45000)
plt.xlim(2000, 2023)

# for exact value
#activete grid
plt.grid()
```

# Scatter plot
- Bivariate Analysis
- numerical Vs numerical
- use case: finding correlation

```python
data = pd.read_csv("C:\\Users\\HP\\Downloads\\batter.csv")
data
```

```python
data = data.head(50)
data
```

```python
plt.scatter(data['avg'], data['strike_rate'])
plt.title('Avg and Strike_rate analysis of top 50 batsman')
plt.xlabel('Avg')
plt.ylabel('Strike_rate')
```

```python
# size
tips = sns.load_dataset('tips')
```

```python
tips
```

```python
# this is slower 
plt.scatter(tips['total_bill'], tips['tip'], s=tips['size']+20)
```

```python
# scatter plot using plt plot
# this is faster technique
plt.plot(tips['total_bill'], tips['tip'], 'o')
```
