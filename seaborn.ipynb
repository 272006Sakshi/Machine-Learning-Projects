## Box Plot
A bosplot is a standardized way of displaying the distribution of data based on a five number summary ('minimum', first quantile(Q1), median, third quantile and maximum). it tells us about outliers and what their values are. it can also tell if the data is symmetrical. how tightly data is grouped, and if and how data is skewed.

- Interquartile Range = (Q3-Q1)
- Minimum = (Q1-1.5*IQR)
- Maximum = (Q3-1.5*IQR)

```python
# univariate analysis
sns.boxplot(tips_data['total_bill'])
```

```python
sns.catplot(x='day', y='total_bill', kind='box', data = tips_data)
```

```python
sns.catplot(x='day', y='total_bill', hue='sex', kind='box', data=tips_data)
```

# Violin Plots
Violin plots are a method of plotting numeric data and can be considered a combination of the box plot with a kernal density plot.

The advantage of the violin plot over the box plot is that aside from showing the statistics it also shows the entire distribution of the data.

Voilin plots are often used to compare the distribution of a given variable across some categories.

```python
sns.violinplot(tips_data['total_bill'])
```

```python
sns.catplot(x='day', y='total_bill', kind='violin', data = tips_data)
```

```python
sns.catplot(x='day', y='total_bill', kind='violin', hue='sex', data = tips_data)
```

```python
sns.catplot(x='day', y='total_bill', kind='violin', hue='sex', split=True, data = tips_data)
```

# Barplot and catplot

```python
# what is the mean value of tip for smoker and non smoker
sns.catplot(x='smoker', y='total_bill', kind='bar', data=tips_data)
```

```python
sns.catplot(x='smoker', y='total_bill', kind='bar', hue='sex', data=tips_data)
```

```python
sns.catplot(x='smoker', y='total_bill', kind='bar', hue='sex',estimator=np.median, data=tips_data)  # we can also use estimator=np.var(for variance), estimator=np.std and all
```

```python
# count plot: it is similar to histogram 
# histogram for numerical data, this is for categorical data

# number of males and females in data
sns.catplot(x='sex', kind='count', data=tips_data)
```

```python
# hue parameter
sns.catplot(x='sex', kind='count', hue = 'smoker', data=tips_data)
```

## Heatmaps
A heatmap is a graphical representation of data where the individual values contained in a matrix are represented as colors.

it's generally drawn between two categorical data sometimes on numerical data but numerical data should be discrete numerical data.

```python
data = sns.load_dataset('flights')
```

```python
data.head()
```

```python
# pivot table
x = data.pivot_table(index='year', columns='month', values='passengers', aggfunc='sum')
x
```

```python
sns.heatmap(x)
```

```python
sns.heatmap(x, cbar=False)
```

```python
sns.heatmap(x, cbar=False, linewidth=0.5)
```

```python
# annot=True, fmt='d'
sns.heatmap(x, cbar=False, linewidth=0.5, annot=True, fmt='d')
```

```python
# for color ->cmap parameter ex: 'summer', 'winter'
sns.heatmap(x, cbar=False, linewidth=0.5, annot=True, fmt='d', cmap='viridis')
```

## Clustermap
 it is same as heatmap, difference is it's cluster the rows and col together on the basis of similarity.

```python
sns.clustermap(x)
```

```python
sns.clustermap(x, metric='correlation')
```

```python
sns.clustermap(x, z_score=0, annot=True, metric='correlation')
```
